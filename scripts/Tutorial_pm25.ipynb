# Tutorial: uso de la función `prediccion()`

Este tutorial muestra cómo usar la función `prediccion()` (definida en `Funcion_pm25.py`) para predecir la concentración de PM2.5 (material particulado, µg/m3) a partir de variables meteorológicas, usando el modelo MLP previamente entrenado y guardado en `main_script.ipynb`.

## Variables requeridas

La función recibe un diccionario (o DataFrame) con las siguientes variables **sin transformar**. La función se encarga internamente de la codificación y el escalado:

| Variable | Descripción | Tipo |
|---|---|---|
| `DEWP` | Punto de rocío (°C) | Numérica |
| `TEMP` | Temperatura (°C) | Numérica |
| `PRES` | Presión atmosférica (hPa) | Numérica |
| `Iws` | Velocidad acumulada del viento (m/s) | Numérica |
| `Is` | Horas acumuladas de nieve | Numérica |
| `Ir` | Horas acumuladas de lluvia | Numérica |
| `cbwd` | Dirección combinada del viento (`'NE'`, `'NW'`, `'SE'`, `'cv'`) | Categórica |

**No es necesario** escalar los valores ni codificar `cbwd` manualmente: `prediccion()` aplica el mismo `scaler` y el mismo orden de columnas que se usaron al entrenar el modelo.

## Ejemplo 1: día frío y con viento fuerte del noroeste

```python
nueva_obs = {
    'DEWP': -10,
    'TEMP': -3,
    'PRES': 1025,
    'Iws': 35.5,
    'Is': 0,
    'Ir': 0,
    'cbwd': 'NW'
}

resultado = prediccion(nueva_obs)
```

## Interpretación del resultado

La función imprime y retorna la predicción de **PM2.5 en µg/m3**, ya en su escala original (internamente el modelo predice en escala logarítmica y `prediccion()` revierte esa transformación automáticamente con `np.expm1`).

Además, clasifica automáticamente el valor según el **Índice de Calidad del Aire (ICAP)** usado en este proyecto:

| Categoría | Rango PM2.5 (µg/m3) |
|---|---|
| Bueno | 0 a 50 |
| Regular (Alerta Temprana) | 51 a 79 |
| Alerta | 80 a 109 |
| Preemergencia | 110 a 169 |
| Emergencia | Más de 170 |
